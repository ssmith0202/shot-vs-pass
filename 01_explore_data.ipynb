## Manual read: Ronaldo, Portugal v Czech Republic, minute 54

**Shot:** open play, xG 0.091, outcome Blocked.
Colours: yellow = shooter, blue = Portugal teammates, red = Czech players. Portugal attacking left to right.

### My judgement (by eye, before any code)

1. **Defenders:** one Czech defender was directly in front of Ronaldo, which is likely the player who blocked the shot. The same defender was also blocking the pass to the nearest teammate.
2. **Teammates:** one teammate was roughly the same distance from goal but slightly wider and in a less crowded position. Other teammates were further from goal but in more space. Pass value is a trade-off between distance to goal and how crowded the receiver is.
3. **Passing lanes:** the only realistic clear lane was to a teammate on the edge of the box, out wide near the corner of the penalty area. That would have been a difficult shot from there.
4. **Verdict:** shoot. All options look low value given the defensive positions, Ronaldo was closest to goal, and a shot here can still produce a rebound, deflection or corner.

### Criteria I used (to formalise in code)

- **Receiver quality:** distance and angle to goal
- **Receiver space:** how many opponents are close to the receiver
- **Lane clearance:** whether an opponent is on or near the line between passer and receiver

Working definition: pass value ≈ P(pass completed) × value of the ball at the receiver's position.

### Notes and limitations

- StatsBomb xG already uses freeze-frame information (defender and goalkeeper positions), so the 0.091 already accounts for the blocking defender. Do not add extra value for "it might get through".
- xG only measures the chance this shot scores. It ignores the value of what follows a blocked or saved shot (rebounds, corners). A pure xG vs pass comparison is therefore biased against shooting close to goal. Possible v2 extension.
- Freeze frames are a snapshot: no player movement direction, speed or time on the ball.
- Freeze frames only include players visible to the broadcast camera, so a missing teammate may be off-screen rather than absent.

```python
import numpy as np

def point_to_segment_distance(p, a, b):
    p, a, b = np.array(p), np.array(a), np.array(b)
    ab = b - a
    t = np.dot(p - a, ab) / np.dot(ab, ab)
    t = np.clip(t, 0, 1)
    closest = a + t * ab
    return np.linalg.norm(p - closest)

# Sanity tests: segment from (0,0) to (10,0)
print(point_to_segment_distance((5, 3), (0, 0), (10, 0)))   # expect 3.0 (beside the middle)
print(point_to_segment_distance((15, 0), (0, 0), (10, 0)))  # expect 5.0 (beyond the receiver)
print(point_to_segment_distance((-2, 0), (0, 0), (10, 0)))  # expect 2.0 (behind the passer)
```

```python
LANE_THRESHOLD = 2.0  # pitch units; an assumption to revisit

passer = shot['location']
teammates = [p for p in shot['shot_freeze_frame'] if p['teammate']]
opponents = [p for p in shot['shot_freeze_frame'] if not p['teammate']]

rows = []
for tm in teammates:
    dists = [point_to_segment_distance(op['location'], passer, tm['location']) for op in opponents]
    min_dist = min(dists)
    blocker = opponents[dists.index(min_dist)]['player']['name']
    rows.append({
        'receiver': tm['player']['name'],
        'receiver_location': tm['location'],
        'closest_opponent_to_lane': blocker,
        'min_distance': round(min_dist, 2),
        'lane_clear': min_dist > LANE_THRESHOLD,
    })

lanes = pd.DataFrame(rows)
lanes
```

```python
def lane_distance(p, a, b):
    """Distance from opponent p to the pass lane a->b.
    Returns infinity if p is behind the passer or beyond the receiver,
    because those players are pressure / receiver marking, not lane blockers."""
    p, a, b = np.array(p), np.array(a), np.array(b)
    ab = b - a
    t = np.dot(p - a, ab) / np.dot(ab, ab)
    if t <= 0 or t >= 1:
        return np.inf
    closest = a + t * ab
    return np.linalg.norm(p - closest)

# Re-test: segment (0,0) to (10,0)
print(lane_distance((5, 3), (0, 0), (10, 0)))   # expect 3.0
print(lane_distance((15, 0), (0, 0), (10, 0)))  # expect inf
print(lane_distance((-2, 0), (0, 0), (10, 0)))  # expect inf
```

```python
LANE_THRESHOLD = 2.0  # pitch units; an assumption to revisit

passer = shot['location']
teammates = [p for p in shot['shot_freeze_frame'] if p['teammate']]
opponents = [p for p in shot['shot_freeze_frame'] if not p['teammate']]

rows = []
for tm in teammates:
    dists = [lane_distance(op['location'], passer, tm['location']) for op in opponents]
    min_dist = min(dists)
    if np.isinf(min_dist):
        blocker = 'none alongside lane'
    else:
        blocker = opponents[dists.index(min_dist)]['player']['name']
    rows.append({
        'receiver': tm['player']['name'],
        'receiver_location': tm['location'],
        'closest_opponent_to_lane': blocker,
        'min_distance': round(min_dist, 2),
        'lane_clear': min_dist > LANE_THRESHOLD,
    })

lanes = pd.DataFrame(rows)
lanes
```

```python
pitch = Pitch(pitch_type='statsbomb', pitch_color='grass', line_color='white')
fig, ax = pitch.draw(figsize=(12, 8))

sx, sy = shot['location']

# Pass lanes: green = clear, red = blocked (at current LANE_THRESHOLD)
for i, row in lanes.iterrows():
    rx, ry = row['receiver_location']
    colour = 'lime' if row['lane_clear'] else 'red'
    pitch.lines(sx, sy, rx, ry, ax=ax, color=colour, lw=2, alpha=0.7, zorder=3)

# Opponents
for op in opponents:
    ox, oy = op['location']
    pitch.scatter(ox, oy, ax=ax, s=200, color='red', edgecolors='white', zorder=4)

# Teammates, labelled with their row number from the lanes table and lane distance
for i, row in lanes.iterrows():
    rx, ry = row['receiver_location']
    pitch.scatter(rx, ry, ax=ax, s=200, color='blue', edgecolors='white', zorder=4)
    pitch.text(rx + 1.5, ry, f"{i}: {row['min_distance']}", ax=ax,
               color='white', fontsize=10, ha='left', va='center', zorder=6,
               bbox=dict(facecolor='black', alpha=0.6, edgecolor='none', pad=2))

# Shooter
pitch.scatter(sx, sy, ax=ax, s=400, color='yellow', edgecolors='black', linewidths=2, zorder=5)

ax.set_title(f"{shot['team']}: {shot['player']} | minute {shot['minute']} | "
             f"xG {shot['shot_statsbomb_xg']:.3f} | lane threshold {LANE_THRESHOLD}")
plt.show()
```

Labelled lane plot confirms my original clear option was Dalot (5), the most robust clear lane (clear up to threshold 2.5). However a defender is right next to Dalot, beyond the end of the lane, so lane clearance overstates him. Bernardo (4) is also ‘clear’ but far from goal.
