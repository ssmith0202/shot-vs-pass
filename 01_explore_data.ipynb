```python
from mplsoccer import Pitch
import matplotlib.pyplot as plt

pitch = Pitch(pitch_type='statsbomb', pitch_color='grass', line_color='white')
fig, ax = pitch.draw(figsize=(12, 8))

# Shooter
shooter_x, shooter_y = shot['location']
pitch.scatter(shooter_x, shooter_y, ax=ax, s=400, color='yellow',
              edgecolors='black', linewidths=2, zorder=5)

# Freeze frame players
for p in shot['shot_freeze_frame']:
    x, y = p['location']
    if p['teammate']:
        pitch.scatter(x, y, ax=ax, s=200, color='blue', edgecolors='white', zorder=4)
    else:
        pitch.scatter(x, y, ax=ax, s=200, color='red', edgecolors='white', zorder=4)

ax.set_title(f"{shot['team']}: {shot['player']} | minute {shot['minute']} | "
             f"xG {shot['shot_statsbomb_xg']:.3f} | {shot['shot_outcome']}")
plt.show()
```

## Manual read: Ronaldo, Portugal v Czech Republic, minute 54

**Shot:** open play, xG 0.091, outcome Blocked.
Colours: yellow = shooter, blue = Portugal teammates, red = Czech players. Portugal attacking left to right.

### My judgement (by eye, before any code)

1. **Defenders:** one Czech defender was directly in front of Ronaldo, which is likely the player who blocked the shot. The same defender was also blocking the pass to the nearest teammate.
2. **Teammates:** one teammate was roughly the same distance from goal but slightly wider and in a less crowded position. Other teammates were further from goal but in more space. Pass value is a trade-off between distance to goal and how crowded the receiver is.
3. **Passing lanes:** the only realistic clear lane was to a teammate on the edge of the box, out wide near the corner of the penalty area. That would have been a difficult shot from there.
4. **Verdict:** shoot. All options look low value given the defensive positions, Ronaldo was closest to goal, and a shot here can still produce a rebound, deflection or corner.

### Criteria I used (to formalise in code)

- **Receiver quality:** distance and angle to goal
- **Receiver space:** how many opponents are close to the receiver
- **Lane clearance:** whether an opponent is on or near the line between passer and receiver

Working definition: pass value ≈ P(pass completed) × value of the ball at the receiver's position.

### Notes and limitations

- StatsBomb xG already uses freeze-frame information (defender and goalkeeper positions), so the 0.091 already accounts for the blocking defender. Do not add extra value for "it might get through".
- xG only measures the chance this shot scores. It ignores the value of what follows a blocked or saved shot (rebounds, corners). A pure xG vs pass comparison is therefore biased against shooting close to goal. Possible v2 extension.
- Freeze frames are a snapshot: no player movement direction, speed or time on the ball.
- Freeze frames only include players visible to the broadcast camera, so a missing teammate may be off-screen rather than absent.
